## 1. The landscape — where this course lives

Python has many plotting libraries, and a first question is simply which ones matter here. This course commits to two: **matplotlib** for the underlying model, and **seaborn** for statistical charts — with a light touch of **plotly** where interactivity genuinely earns its place. Bokeh, Altair, and the others are capable tools, but they sit outside this scope, and interactive dashboards belong to a separate course. The map below marks the territory.

## 2. The mental model — Figure, Axes, Axis

Almost every chart in Python is drawn by matplotlib — directly, or under seaborn. One small object hierarchy stands behind all of it, and settling it early turns much of the rest of the course into vocabulary. The first diagram names the parts of a single figure, kept to only the pieces this course actually uses.

Those parts nest. A **Figure** is the whole canvas; it holds one or more **Axes**, each a single plot; and each Axes carries an **Axis** for x and for y — the number lines. The same structure extends to a grid of plots inside one Figure.

> **Note:** An `Axes` is a whole plot, not a single line. That one naming trap accounts for most early confusion — a Figure is the canvas, an Axes is one plot on it, and an Axis is the x or y number line inside that plot.

## 3. How this course works

A good chart is a series of deliberate choices, and the default is rarely the best one. That idea runs through every lesson: each opens with a real question and answers it with a chart, rather than touring an API. Teaching runs on a cleaned dataset of roughly 100,000 hospital encounters, joined at times with county health data and a daily respiratory time series. A separate project — rural hospital closures — is reserved for the capstone and carried across both sessions. Interactivity stays deliberately small: three sliders and one hover chart, nothing that drifts toward a dashboard.

> **Tip:** Every lesson ships clean, with no stored output. Run the cells top to bottom, work each exercise in its empty cell first, then expand the collapsed solution to check.

## 4. The roadmap

The course spans two sessions of about three and a half hours each. Session 1 builds from a single variable up to comparisons across groups; Session 2 moves to relationships, time, and making a figure ready for an audience. The rural-hospital capstone lands in two parts — once after categories, and again at the very end, where the whole toolkit comes together on one presentation-ready figure.
