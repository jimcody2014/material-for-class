## 1. Why NumPy

An array looks like a list, but it is **typed**, stored compactly, and supports **vectorized** operations — whole-array math with no explicit loop. On real datasets this is also dramatically faster.

````python
import numpy as np

counts = np.array([120, 95, 130, 88, 145])
print(counts, "| dtype:", counts.dtype)
````

````python
# Python list: you loop (or comprehend) to scale every value
py = [120, 95, 130]
print([c * 1.1 for c in py])

# NumPy: one vectorized expression — no loop
arr = np.array([120, 95, 130])
print(arr * 1.1)
````

> Reference: [NumPy quickstart](https://numpy.org/doc/stable/user/quickstart.html).

## 2. Creating arrays

From a list, or with constructors for ranges, evenly-spaced points, constants, and random values.

````python
# np.array([1, 2, 3])
# np.arange(0, 10, 2)      # start, stop (exclusive), step
# np.linspace(0, 1, 5)     # 5 evenly spaced points, endpoints included
# np.zeros(3)
# np.ones((2, 3))          # shape given as a tuple
````

````python
# Modern reproducible random numbers
rng = np.random.default_rng(seed=0)
print(rng.integers(0, 100, size=5))             # 5 ints in [0, 100)
print(rng.normal(loc=70, scale=10, size=5))     # e.g. simulated resting heart rates
````

Every array has a single `dtype`. Mixing an integer and a float promotes the whole array to float.

````python
a = np.array([1, 2, 3])        # integers
b = np.array([1.0, 2, 3])      # one float -> all float
print(a.dtype, "|", b.dtype)
````

## 3. Vectorized operations

Arithmetic and comparisons apply **element-wise**. This is how you'll express most calculations — no loops.

````python
weights = np.array([72.5, 80.1, 65.3, 90.0])   # kg
heights = np.array([1.70, 1.82, 1.60, 1.88])   # m

bmi = weights / heights**2                     # element-wise across both arrays
print(np.round(bmi, 1))
````

````python
# Comparisons return a boolean array (remember this for §5)
print(bmi > 25)
````

## 4. Aggregation along an axis

Reductions like `sum` and `mean` can collapse the whole array or run **along one axis**. For a 2D array: `axis=0` collapses **down the rows** (a result per column), `axis=1` collapses **across the columns** (a result per row). This maps directly onto `df.sum(axis=...)` in pandas.

````python
# rows = clinics, columns = weekday counts (Mon..Fri)
week = np.array([[120,  95, 130,  88, 145],
                 [ 60,  72,  65,  80,  70],
                 [200, 180, 210, 190, 205]])

print("grand total :", week.sum())
print("per clinic   :", week.sum(axis=1))   # across the 5 weekdays
print("per weekday  :", week.mean(axis=0))  # across the 3 clinics
print("busiest day  :", week.max(axis=1))   # peak day per clinic
````

## 5. Boolean masking & indexing

A comparison gives a boolean array; using it to index keeps only the `True` positions. This exact pattern becomes the way you filter rows in pandas.

````python
labs = np.array([5.1, 7.4, 9.2, 6.8, 8.0, 5.5])

mask = labs >= 7.0
print(mask)
print(labs[mask])           # keep values where the mask is True
print(labs[labs >= 7.0])    # the same thing in one step
print((labs >= 7.0).sum())  # how many qualify (True counts as 1)
````

Indexing a 2D array uses `[row, column]`; a bare `:` means *all*.

````python
# week[0]      # first row  -> first clinic
# week[:, 0]   # first column -> Monday across all clinics
# week[2, 4]   # row 2, column 4
````

## 6. Shape & reshape

`.shape` reports the dimensions; `.reshape` rearranges the same data into a new shape (the element count must match).

````python
a = np.arange(12)
print(a.shape)          # (12,)

b = a.reshape(3, 4)     # 3 rows, 4 columns
print(b)
print(b.shape)          # (3, 4)
````

## 7. Broadcasting

NumPy can combine arrays of different shapes by "stretching" the smaller one to fit — so you can, for example, apply a per-column adjustment to every row without a loop.

````python
counts = np.array([[120, 95, 130],
                   [ 60, 72,  65]])
adjust = np.array([10, 0, -5])      # one value per column

print(counts + adjust)              # 'adjust' is broadcast across both rows
````

> Reference: [Broadcasting rules](https://numpy.org/doc/stable/user/basics.broadcasting.html).
