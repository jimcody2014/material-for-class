The same data can be laid out **wide** (a column per measure) or **long/tidy** (one row per observation). Most grouping and plotting wants long format; reports often want wide. These tools convert between them:

- **`melt`** — wide → long
- **`pivot`** / **`pivot_table`** — long → wide
- **`stack`** / **`unstack`** — move columns to/from the index

## 1. Wide vs. long

Here's the same clinic data in **wide** form — one column per month. Easy to read, but awkward to group or plot.

```python
wide = pd.DataFrame({
    "site": ["Riverside", "Hilltop", "Downtown"],
    "Jan":  [4200, 2600, 1500],
    "Feb":  [3800, 2400, 1400],
    "Mar":  [4100, 2500, 1600],
})
wide
```

## 2. `melt` — wide → long

`melt` unpivots the month columns into rows. `id_vars` are the columns to **keep** as identifiers; everything else collapses into a name column and a value column.

```python
long = wide.melt(id_vars="site", var_name="month", value_name="visits")
long
```

Now each row is one observation (site, month, visits) — the tidy shape that `groupby` and most plotting libraries expect.

```python
long.groupby("month")["visits"].sum()   # trivial once the data is long
```

> Reference: [Reshaping and pivot tables](https://pandas.pydata.org/docs/user_guide/reshaping.html).

## 3. `pivot` / `pivot_table` — long → wide

`pivot` is the inverse of `melt`: choose what becomes the index, the columns, and the values.

```python
long.pivot(index="site", columns="month", values="visits")
```

`pivot` errors if an index/column pair repeats. `pivot_table` handles that by **aggregating** (it's a spreadsheet pivot table) — useful when there are duplicates to summarize.

```python
long.pivot_table(index="site", columns="month", values="visits", aggfunc="mean")
```

## 4. `stack` / `unstack`

`stack` pushes columns down into the index (producing a long Series); `unstack` pulls an index level back out into columns. They're the index-based cousins of melt/pivot.

```python
wide_i = wide.set_index("site")
stacked = wide_i.stack()      # columns (months) move into the index
stacked
```

```python
stacked.unstack()             # and back to wide
```
