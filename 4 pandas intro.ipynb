## 1. Series

A one-dimensional labeled array. Build one from a list (with an optional index) or straight from a dict.

````python
visits = pd.Series([218, 201, 216, 132], index=["Mon", "Tue", "Wed", "Thu"])
visits
````

````python
# visits["Wed"]        # access by label
# visits.iloc[0]       # access by position
# visits[visits > 200] # boolean mask — same idea as NumPy
````

````python
# A dict becomes a Series with the keys as the index
pop = pd.Series({"Riverside": 42000, "Hilltop": 18500, "Downtown": 9300})
pop
````

> Reference: [Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html).

## 2. Index alignment

This is the behavior that surprises people coming from NumPy or other languages: arithmetic between Series matches on the **index label**, not on position. Labels present in only one side produce `NaN`.

````python
a = pd.Series({"Riverside": 100, "Hilltop": 50})
b = pd.Series({"Hilltop": 5, "Riverside": 10})   # note: different order

a + b    # matched by label (Riverside+Riverside, Hilltop+Hilltop), not by position
````

````python
# A label on only one side -> NaN
c = pd.Series({"Riverside": 1, "Downtown": 9})
a + c
````

## 3. DataFrames

A DataFrame is the workhorse. The most common way to build one by hand is from a dict of columns.

````python
data = {
    "site":   ["Riverside", "Hilltop", "Downtown"],
    "visits": [218, 132, 76],
    "wait":   [32.7, 21.4, 15.9],
}
df = pd.DataFrame(data)
df
````

## 5. Inspecting & selecting

`.info()` is the fastest way to see column names, non-null counts, and dtypes at a glance.

````python
diabetes.info()
````

Select **one column with single brackets** (you get a Series) and **multiple columns with a list inside the brackets** (you get a DataFrame).

````python
age = diabetes["age"]                 # single bracket -> Series
print(type(age))

subset = diabetes[["age", "gender"]]  # list of names -> DataFrame
print(type(subset))
subset.head()
````

Rows come out by label with `.loc` and by position with `.iloc`. We'll go deep on these in the **Indexes** lesson; for now, the basics:

````python
print(df.loc[0])      # row by index label
print(df.iloc[0])     # row by position
df.iloc[0:2]          # first two rows
````

## 6. Sorting

Order rows by a column's values, by the index, or grab the top/bottom N directly.

````python
df.sort_values("visits", ascending=False)   # rows ordered by visits, highest first
````

````python
df.nlargest(2, "visits")                     # shortcut for the top 2 by visits
````

> `sort_index()` orders by the index instead of a column; `nsmallest` is the bottom-N counterpart.
